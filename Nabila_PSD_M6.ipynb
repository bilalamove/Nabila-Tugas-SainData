```python
import pandas as pd
import numpy as np
from IPython.display import display

df = pd.read_csv("paket_a_transaksi_pos_2024_2025.csv")

print("Ukuran dataset:", df.shape)
display(df.head())
df.info()
```

**Task 1 Indexing & Selection**

```python
# Membersihkan format harga
print("Tipe data sebelum dibersihkan:", df["harga_satuan"].dtype)
display(df["harga_satuan"].head(10))

df["harga_satuan"] = pd.to_numeric(
    df["harga_satuan"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
).astype(float)

print("Tipe data setelah dibersihkan:", df["harga_satuan"].dtype)
display(df["harga_satuan"].head(10))
```

```python
# Penanganan kuantitas negatif
jumlah_negatif = (df["jumlah"] < 0).sum()
print("Jumlah nilai jumlah yang negatif:", jumlah_negatif)

df.loc[df["jumlah"] < 0, "jumlah"] = df.loc[
    df["jumlah"] < 0, "jumlah"
].abs()

print("Jumlah nilai negatif setelah diperbaiki:", (df["jumlah"] < 0).sum())
```

```python
# Imputasi nilai kosong kasir
print("NaN kode_kasir sebelum imputasi:", df["kode_kasir"].isna().sum())

df.loc[df["kode_kasir"].isna(), "kode_kasir"] = "K_UNKNOWN"

print("NaN kode_kasir setelah imputasi:", df["kode_kasir"].isna().sum())
```

```python
# Seleksi posisi slicing
display(df.iloc[:5, [0, 1, 2, 5]])
```

**Task 2 Operasi Aritmatika**

```python
# Perhitungan Omzet kotor
df["Total_Kotor"] = df["harga_satuan"] * df["jumlah"]

display(df[["harga_satuan", "jumlah", "Total_Kotor"]].head())
```

```python
# Perhitungan Diskon
df["Nominal_Diskon"] = df["Total_Kotor"] * (df["diskon_persen"] / 100)

display(df[["Total_Kotor", "diskon_persen", "Nominal_Diskon"]].head())
```

```python
# Perhitubngan omzet bersih dan pajak
df["Total_Bersih"] = (
    df["Total_Kotor"] - df["Nominal_Diskon"]
) * 1.11

display(df[[
    "Total_Kotor", "Nominal_Diskon", "Total_Bersih"
]].head())
```

**Task 3 Filtering**

```python
# Filter multi kondisi
filter_multi = df.loc[
    (df["status"] == "SELESAI") &
    (df["kode_gerai"].isin(["G05", "G09"])) &
    (df["Total_Bersih"] > 100_000)
]

print("Jumlah transaksi yang memenuhi kriteria:", len(filter_multi))
display(filter_multi)
```

```python
# Filter dengan query
filter_g10 = df.query("kode_gerai == 'G10' and jumlah >= 5")

print("Jumlah transaksi G10 dengan jumlah minimal 5 unit:", len(filter_g10))
display(filter_g10)
```

**Task 4 Sorting**

```python
# Multi colomn sorting
df_sorted = df.sort_values(
    by=["kode_gerai", "Total_Bersih"],
    ascending=[True, False]
)

display(df_sorted)
```

```python
# Top N records
top_5_transaksi = df.nlargest(5, "Total_Bersih")

display(top_5_transaksi[[
    "id_transaksi", "tanggal", "kode_gerai",
    "jumlah", "Total_Bersih"
]])
```

**Task 5 Grouping dan agregasi**

```python
# Laporan agregasi per gerai
df_selesai = df.loc[df["status"] == "SELESAI"].copy()

laporan_gerai = (
    df_selesai.groupby("kode_gerai")
    .agg(
        Jumlah_Transaksi=("id_transaksi", "count"),
        Total_Unit_Terjual=("jumlah", "sum"),
        Total_Omzet_Bersih=("Total_Bersih", "sum"),
        Rata_Rata_Transaksi=("Total_Bersih", "mean")
    )
    .sort_values("Total_Omzet_Bersih", ascending=False)
    .round(2)
)

display(laporan_gerai)
```

```python
# Transformasi kelompok
total_omzet_per_gerai = (
    df.groupby("kode_gerai")["Total_Bersih"].transform("sum")
)

df["Kontribusi_Gerai_Persen"] = (
    df["Total_Bersih"] / total_omzet_per_gerai) * 100

display(df[[
    "id_transaksi", "kode_gerai", "status",
    "Total_Bersih", "Kontribusi_Gerai_Persen"
]].head(10))
```
